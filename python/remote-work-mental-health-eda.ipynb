```python
# Correlation matrix heatmap with only numeric columns
plt.figure(figsize=(12, 8))
numeric_df = df.select_dtypes(include=[np.number])  # Select only numeric columns
correlation_matrix = numeric_df.corr()

sns.heatmap(correlation_matrix, annot=True, fmt='.2f', cmap='coolwarm', mask=np.triu(correlation_matrix))
plt.title('Correlation Matrix')
plt.show()
```

# <div style="text-align:center; border-radius:15px 50px; padding:15px; color:white; margin:0; font-size:100%; font-family:Arial; background-color:purple; overflow:hidden"><b> ✨ 3D Scatter Plot ✨</b></div>

```python
# Cell 15: Import additional libraries for styling and advanced visuals
import matplotlib.dates as mdates

# Cell 16: Violin plot for satisfaction levels
plt.figure(figsize=(12, 6))
sns.violinplot(x='Satisfaction_with_Remote_Work', data=df, palette='muted')
plt.title('Violin Plot of Satisfaction with Remote Work')
plt.xlabel('Satisfaction Level')
plt.ylabel('Density')
plt.grid(axis='y')
plt.show()

# Cell 17: Scatter plot for Age vs. Hours Worked
plt.figure(figsize=(12, 6))
sns.scatterplot(x='Age', y='Hours_Worked_Per_Week', hue='Satisfaction_with_Remote_Work', data=df, palette='deep')
plt.title('Scatter Plot of Age vs. Hours Worked Per Week')
plt.xlabel('Age')
plt.ylabel('Hours Worked Per Week')
plt.legend(title='Satisfaction Level')
plt.show()

# Cell 18: Pairplot to visualize relationships between features
sns.pairplot(df, hue='Satisfaction_with_Remote_Work', palette='coolwarm')
plt.title('Pairplot of Features by Satisfaction Level')
plt.show()

# Cell 19: Trend line for Hours Worked vs. Age
plt.figure(figsize=(12, 6))
sns.regplot(x='Age', y='Hours_Worked_Per_Week', data=df, scatter_kws={'alpha':0.5}, line_kws={'color':'red'})
plt.title('Trend Line for Hours Worked Per Week vs. Age')
plt.xlabel('Age')
plt.ylabel('Hours Worked Per Week')
plt.show()

# Cell 20: Count of each satisfaction level as a pie chart
satisfaction_counts = df['Satisfaction_with_Remote_Work'].value_counts()
plt.figure(figsize=(8, 8))
plt.pie(satisfaction_counts, labels=satisfaction_counts.index, autopct='%1.1f%%', startangle=140, colors=sns.color_palette('pastel'))
plt.title('Pie Chart of Satisfaction Levels')
plt.axis('equal')
plt.show()

# Cell 21: Count plot with percentage display
plt.figure(figsize=(12, 6))
sns.countplot(data=df, x='Satisfaction_with_Remote_Work', palette='viridis')
total = len(df)
for p in plt.gca().patches:
    plt.annotate(f'{p.get_height()/total:.2%}', (p.get_x() + p.get_width() / 2., p.get_height()), 
                 ha='center', va='bottom', fontsize=12)
plt.title('Count Plot of Satisfaction with Percentage')
plt.xlabel('Satisfaction Level')
plt.ylabel('Count')
plt.show()

# Cell 22: Boxen plot for age distribution
plt.figure(figsize=(12, 6))
sns.boxenplot(x='Satisfaction_with_Remote_Work', y='Age', data=df, palette='Set2')
plt.title('Boxen Plot of Age by Satisfaction Level')
plt.xlabel('Satisfaction Level')
plt.ylabel('Age')
plt.show()

# Cell 23: Time series analysis (if applicable)
# Assuming you have a date column; Uncomment and adjust accordingly
# df['Date'] = pd.to_datetime(df['Date'])
# df.set_index('Date', inplace=True)
# plt.figure(figsize=(12, 6))
# df['Satisfaction_with_Remote_Work'].resample('M').mean().plot()
# plt.title('Monthly Average Satisfaction with Remote Work')
# plt.xlabel('Date')
# plt.ylabel('Average Satisfaction')
# plt.show()

# Cell 24: Custom color palette for visualizations
custom_palette = sns.color_palette("dark:.3")

```

```python
# Cell 26: 3D scatter plot of Age, Hours Worked and Satisfaction Level
# Satisfaction is ordinal. It is mapped to 1 (Unsatisfied), 2 (Neutral), 3 (Satisfied) for plotting only.
import matplotlib.pyplot as plt
from mpl_toolkits.mplot3d import Axes3D

satisfaction_order = {'Unsatisfied': 1, 'Neutral': 2, 'Satisfied': 3}
sat_num = df['Satisfaction_with_Remote_Work'].map(satisfaction_order)

fig = plt.figure(figsize=(12, 8))
ax = fig.add_subplot(111, projection='3d')
scatter = ax.scatter(df['Age'], df['Hours_Worked_Per_Week'], sat_num,
                     c=sat_num, cmap='plasma', s=50)

ax.set_xlabel('Age')
ax.set_ylabel('Hours Worked Per Week')
ax.set_zlabel('Satisfaction Level (1 = Unsatisfied, 3 = Satisfied)')
ax.set_zticks([1, 2, 3])
ax.set_title('3D Scatter Plot: Age, Hours Worked, and Satisfaction Level')

cbar = plt.colorbar(scatter)
cbar.set_label('Satisfaction Level (1 to 3)')
cbar.set_ticks([1, 2, 3])
plt.show()
```
